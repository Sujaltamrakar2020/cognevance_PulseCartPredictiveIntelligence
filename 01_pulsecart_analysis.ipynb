# PulseCart — Beginner Analytics Notebook
This notebook mirrors the main analysis in simple steps.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/processed/pulsecart_clean.csv')
df.head()
```

```python
df.describe(numeric_only=True)
```

```python
df.groupby('region')['revenue_90d'].sum().sort_values(ascending=False)
```

```python
df.groupby('membership')['repeat_purchase_30d'].mean().mul(100).round(1)
```

```python
df.plot.scatter(x='days_since_last_purchase', y='revenue_90d', alpha=.35)
plt.title('Recency vs Revenue')
plt.show()
```
